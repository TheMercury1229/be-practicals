### Analyzing Customer Churn in a Telecommunications Company

- dataset link:https://raw.githubusercontent.com/harshbg/Telecom-Churn-Data-Analysis/master/Telecom%20Churn.csv

```python
#Import the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Load the datasets
df = pd.read_csv("./datasets/assignment-06.csv")

print("Training dataset shape:", df.shape)

display(df.head())
```

```python
# Explore the dataset
print("Training dataset information:")
df.info()

print("Statistical summary of the training dataset:")
display(df.describe(include='all'))

print("Missing values in the training dataset:")
print(df.isnull().sum())

print("\nDuplicate records:")
print("Training:", df.duplicated().sum())
```

```python
# Handle missing values
df = df.dropna(subset=['churn']).copy()

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(include=["object","string"]).columns.tolist()

for col in numeric_cols:
    df[col]=df[col].fillna(df[col].median())

for col in categorical_cols:
    if df[col].isnull().any():
        df[col]=df[col].fillna(df[col].mode()[0])


print("Remaining missing values",df.isnull().sum())
```

```python
# Remove duplicate records

df = df.drop_duplicates().reset_index(drop=True)
print("Duplicates after cleaning:",df.duplicated().sum())
```

```python
# Remove the inconsistent data
categorical_columns=df.select_dtypes(include=["object","string"]).columns
for col in categorical_columns:
    df[col]=df[col].astype('string').str.strip()

for col in ['International plan','Voice mail plan']:
    if col in df.columns:
        df=df[col].str.lower().map({"yes":"Yes","no":"No"}).fillna(df[col])


if "State" in df.columns:
    df=df["State"].str.upper()
```

```python
# Convert columns to appropriate data types

#convert numeric columns
for col in numeric_cols:
    df[col]=pd.to_numeric(df[col],errors='coerce')


# Handle any invalid numerical values 
for col in numeric_cols:
    df[col]=df[col].fillna(df[col].median())

# Standardize churn target
df['churn'] = (
    df["churn"].astype(str).str.strip().str.lower()
    .map({"true": True, "false": False, "yes": True, "no": False,
          "1": True, "0": False})
)

df = df.dropna(subset=['churn']).reset_index(drop=True)


print("Updated data types:")
print(df.dtypes)
```

```python
# Detect and handle outliers

outlier_cols = [
    col for col in df.select_dtypes(include=np.number).columns
    if col not in ["area code","churn"]
]

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col]<lower)|(df[col]>upper)).sum()
    print(f"{col}:{count} outliers deteted")

    df[col]=df[col].clip(lower,upper)

print("Outlier handling completed.")
```

```python
# Perform feature engineering

df["total minutes"] = (
    df["total day minutes"]
    + df["total eve minutes"]
    + df["total night minutes"]
    + df["total intl minutes"]
)

df["total calls"] = (
    df["total day calls"]
    + df["total eve calls"]
    + df["total night calls"]
    + df["total intl calls"]
)

df["total charges"] = (
    df["total day charge"]
    + df["total eve charge"]
    + df["total night charge"]
    + df["total intl charge"]
)

display(df[["total minutes", "total calls", "total charges", "churn"]].head())
```

```python
# Scale features
excluded_cols= ['churn','state','international plan','voice main plan','area code']

scale_cols=[col for col in df.select_dtypes(include=np.number).columns if col not in excluded_cols]


scaler = StandardScaler()
df[scale_cols] = scaler.fit_transform(df[scale_cols])

print("Numerical features scaled.")
display(df[scale_cols].head())
```

```python
# Spilt in train and test

X = df.drop(columns=["churn"])
y = df["churn"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training targets:", y_train.shape)
print("Testing targets:", y_test.shape)
```

```python
# Export to clean csv
df.to_csv("./outputs/assingment-06-op.csv",index=False)

# Export the train and test datasets
X_train.to_csv("./outputs/assingment-06-op-train.csv", index=False)
X_test.to_csv("./outputs/assingment-06-op-test.csv", index=False)
y_train.to_csv("./outputs/assingment-06-op-train-targets.csv", index=False)
y_test.to_csv("./outputs/assingment-06-op-test-targets.csv", index=False)


print("Cleaned dataset and train-test splits exported successfully.")
```
