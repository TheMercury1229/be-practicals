### Analyzing Sales Data from Multiple File Formats

- Dataset link:https://www.kaggle.com/datasets/jr2ngb/superstore-data

```python
# Import the libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the original csv and create excel and json files
df_original = pd.read_csv("./datasets/assignment-02.csv",encoding="cp1252")

# Create equivalent excel and json files
df_original.to_excel("./datasets/assignment-02.xlsx", index=False)
df_original.to_json("./datasets/assignment-02.json", orient="records", date_format="iso")

print("CSV, Excel, and JSON files have been created successfully.")
print("Original dataset shape",df_original.shape)

display(df_original.head())
```

```python
# Load the data from all three formats
csv_df = pd.read_csv("./datasets/assignment-02.csv",encoding="cp1252")
excel_df = pd.read_excel("./datasets/assignment-02.xlsx")
json_df = pd.read_json("./datasets/assignment-02.json")

print("CSV DataFrame shape:", csv_df.shape)
print("Excel DataFrame shape:", excel_df.shape)
print("JSON DataFrame shape:", json_df.shape)


display(csv_df.head())
display(excel_df.head())
display(json_df.head())
```

```python
# Explore structure and data quality of the dataset
print("Column names:")
print(csv_df.columns.tolist())

print("\nData types:")
print(csv_df.dtypes)

print("\nMissing values:")
print(csv_df.isnull().sum())

print("\nDuplicate rows:", csv_df.duplicated().sum())

print("\nStatistical summary:")
display(csv_df.describe(include="all").T)
```

```python
# Clean the data 
def clean_data(data):
    data = data.copy()

    data.columns = (
        data.columns.str.strip()
        .str.lower()
        .str.replace(r"[^a-z0-9]+", "_", regex=True)
        .str.strip("_")
    )

    data = data.drop_duplicates()
    # Correct data types
    for col in ["order_date", "ship_date"]:
        if col in data.columns:
            data[col] = pd.to_datetime(data[col], errors="coerce")

    for col in data.select_dtypes(include="number").columns:
        data[col]=data[col].fillna(data[col].median())

    for col in data.select_dtypes(include="object").columns:
        if data[col].isnull().any():
            mode_value = data[col].mode()
            if not mode_value.empty:
                data[col] = data[col].fillna(mode_value[0])

    return data


csv_clean=clean_data(csv_df)
excel_clean=clean_data(excel_df)
json_clean=clean_data(json_df)


print("CSV missing values:", csv_clean.isnull().sum().sum())
print("Excel missing values:", excel_clean.isnull().sum().sum())
print("JSON missing values:", json_clean.isnull().sum().sum())
```

```python
# Convert to a unified dataframe
df = pd.concat([csv_clean, excel_clean, json_clean], ignore_index=True)

print("Unified DataFrame shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
# verify the consistency between formats 
print("CSV and Excel have matching shapes:",
      csv_clean.shape == excel_clean.shape)

print("CSV and JSON have matching shapes:",
      csv_clean.shape == json_clean.shape)

print("CSV and Excel contain matching records:",
      csv_clean.equals(excel_clean))

print("CSV and JSON contain matching records:",
      csv_clean.equals(json_clean))
```

```python
# Transform the data for analysis
df["order_year"]=df["order_date"].dt.year
df["order_month"]=df["order_date"].dt.month

df["profit_margin"]=(df["profit"]/df["sales"].replace(0,pd.NA))*100


# Split product name into first word and remaining description
df["product_first_word"] = df["product_name"].str.split().str[0]

display(df[
    ["order_id", "order_date", "order_year", "order_month",
     "product_name", "product_first_word", "sales", "profit_margin"]
].head())
```

```python
# Merge sales with product details 
product_details = (
    df[["product_id","product_name","category","sub_category"]]
    .drop_duplicates(subset="product_id")
)

sales_merged = pd.merge(
    df,
    product_details,
    on="product_id",
    how="left",
    suffixes=('',"_lookup")
)

display(sales_merged.head())
```

```python
# Calculate key sales metrics
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_quantity = df["quantity"].sum()
total_orders = df["order_id"].nunique()
avg_order_value = df.groupby("order_id")["sales"].sum().mean()


print("Total sales:", round(total_sales, 2))
print("Total profit:", round(total_profit, 2))
print("Total quantity sold:", total_quantity)
print("Number of unique orders:", total_orders)
print("Average order value:", round(avg_order_value, 2))
```

```python
# Sales by product category
category_summary = (
    df.groupby("category")
    .agg(
        total_sales=("sales","sum"),
        total_profit=("profit","sum"),
        total_quantity=("quantity","sum"),
    )
    .reset_index()
    .sort_values("total_sales",ascending=False)
)

display(category_summary)
```

```python
# Sales by region
region_summary = (
    df.groupby("region")
    .agg(
        total_sales=("sales","sum"),
        total_profit=("profit","sum"),
    )
    .reset_index()
    .sort_values("total_sales",ascending=False)
)

display(region_summary)
```

```python
# Monthly sales trend
monthly_sales =(
    df.groupby(["order_year","order_month"])["sales"]
    .sum()
    .reset_index()
)

monthly_sales["period"]=(
    monthly_sales["order_year"].astype(str)+"-"
    +monthly_sales["order_month"].astype(str).str.zfill(2)
)

plt.figure(figsize=(12,5))
sns.lineplot(data=monthly_sales,x="period",y="sales",marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Sales by category
plt.figure(figsize=(8, 5))
sns.barplot(data=category_summary, x="category", y="total_sales")
plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
# Category distribution
plt.figure(figsize=(7, 7))
plt.pie(
    category_summary["total_sales"],
    labels=category_summary["category"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Share of Total Sales by Category")
plt.show()
```

```python
# Box plot:sales distribution
plt.figure(figsize=(8, 5))
sns.boxplot(y=df["sales"])
plt.title("Distribution of Sales Values")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
# Final results and save cleaned data
print("Final dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

display(df.head())

df.to_csv("./outputs/assignment-02.csv", index=False)
print("Cleaned sales data saved.")
```
