### Data Aggregation

```python
# Import the required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_csv("./datasets/assignment-12.csv", encoding="latin1")
df.head()
```

```python
# Explore the dataset
# Display the first five rows
display(df.head())

# Display the number of rows and columns
print("Dataset shape:", df.shape)

# Display column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset information:")
df.info()

# Check for missing values
print("\nMissing values:")
print(df.isnull().sum())

# Display summary statistics
print("\nStatistical summary:")
display(df.describe())
```

```python
# Select the relevant variables
sales_data=df[['Region','Category','Sales']].copy()

# Display the selected data
display(sales_data.head())

# Check data types
print(sales_data.dtypes)

print("Missing values in the selected data:")
print(sales_data.isnull().sum())
```

```python
# Calculate total sales by region
region_sales=sales_data.groupby('Region')['Sales'].sum()

region_sales=region_sales.sort_values(ascending=False)

print("Total Sales by Region:")
display(region_sales)
```

```python
# Visualize sales distribution by region

plt.figure(figsize=(8,5))
region_sales.plot(kind='bar',edgecolor="black")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y",linestyle="--",alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# OPTIONAL:Pie chart
plt.figure(figsize=(7,7))
plt.pie(region_sales, labels=region_sales.index,autopct="%1.1f%%",startangle=90)

plt.title("Sales distribution by Region")
plt.show()
```

```python
# Idetify the top performing region
top_region =region_sales.idxmax()
top_sales = region_sales.max()

print("Top performing region:", top_region)
print("Total sales:",round(top_sales,2))

print("\nRegional sales ranking:")
display(region_sales)
```

```python
# Aggregate sales by region and product category
region_category_sales = sales_data.groupby(
    ["Region","Category"]
)["Sales"].sum().unstack()

print("Sales by region and product category:")
display(region_category_sales)
```

```python
# Create a stacked bar chart
region_category_sales.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6),
    edgecolor="black"
)

plt.title("Sales by Region and Product Category")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.legend(title="Product Category")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# Create a grouped bar chart

region_category_sales.plot(
    kind="bar",
    stacked=False,
    figsize=(10, 6),
    edgecolor="black"
)

plt.title("Comparison of Category Sales Across Regions")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.legend(title="Product Category")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
#FINAL SUMMARY OF DATA AGGREGATION
print("DATA AGGREGATION SUMMARY")
print("-" * 35)

print("\nTotal sales across all regions:",
      round(sales_data["Sales"].sum(), 2))

print("\nSales by region:")
display(region_sales)
print("\nSales by region and category:")
display(region_category_sales)
```
