### Implement a program for retrieval of documents using inverted files.

```python
#Define the documents
documents = {
    "D1": "Python is easy to learn",
    "D2": "Python is used for machine learning",
    "D3": "Information retrieval uses an inverted index",
    "D4": "Python is useful for information retrieval"
}

print("Documents:")
for doc_id, text in documents.items():
    print(doc_id, ":", text)
```

```python
# Create an inverted index
import re
inverted_index = {}
for doc_id, text in documents.items():
    # Tokenize the text into words
    words = re.findall(r'\b\w+\b', text.lower()) 

    for word in words:
        if word not in inverted_index:
            inverted_index[word]=[]
        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)

print("\nInverted Index:\n")
for word, doc_ids in sorted(inverted_index.items()):
    print(word, ":", doc_ids)
```

```python
# Search for a document

query = input("Enter a search word: ").lower().strip()

if query in inverted_index:
    print("Documents containing",query,":")
    for doc_id in inverted_index[query]:
        print(doc_id, ":", documents[doc_id])
else:
    print("No documents found containing")
```

```python
# Support multiple words
query = input("Enter search words: ").lower().split()

if query:
    result_sets = [
        set(inverted_index.get(word, []))
        for word in query
    ]

    matching_docs = set.intersection(*result_sets)

    if matching_docs:
        print("Matching Documents:")
        for doc_id in sorted(matching_docs):
            print(doc_id, ":", documents[doc_id])
    else:
        print("No matching documents found.")
else:
    print("Please enter a search query.")
```
